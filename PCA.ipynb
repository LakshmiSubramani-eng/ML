```python
import pandas as pd
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.decomposition import PCA
from sklearn.preprocessing import MinMaxScaler
```

```python
df=load_breast_cancer()
```

```python
x=pd.DataFrame(data=df.data,columns=df.feature_names)
```

```python
x
```

```python
y=pd.DataFrame(df["target"])
```

```python
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,train_size=0.80,random_state=10)
```

```python
model=DecisionTreeClassifier()
```

```python
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_train)
```

```python
accuracy_score(y_train,y_pred)
```

```python
y_test_predict=model.predict(x_test)
```

```python
accuracy_score(y_test,y_test_predict)
```

```python
min_max_scaler=MinMaxScaler()
```

```python
scaler=min_max_scaler.fit_transform(x)
```

```python
scaler
```

```python
scaler_df=pd.DataFrame(data=scaler,columns=df.feature_names)
```

```python
scaler_df
```

```python
pca=PCA(n_components=3)
```

```python
x_pca=pca.fit_transform(scaler_df)
```

```python
x_pca=pd.DataFrame(data=x_pca)
```

```python
x_pca
```

```python
x_train,x_test,y_train,y_test=train_test_split(x_pca,y,train_size=0.80,random_state=10)
model=DecisionTreeClassifier()
model.fit(x_train,y_train)
y_pred=model.predict(x_train)
print(accuracy_score(y_train,y_pred))
y_test_predict=model.predict(x_test)
print(accuracy_score(y_test,y_test_predict))
```
