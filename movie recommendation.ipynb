```python
def movie_recom(name,n): 
    dis=linear_kernel(transformdata[indexed[name]],transformdata)
    score=pd.DataFrame(dis).T
    score.columns=["score"]
    score.sort_values("score",ascending=False)
    lst=[]
    for i in range(0,n+1):
        Movie_title={"Movie_title":movie_df['Movie Title'][score.index[i]],
                     "Ratinng":movie_df['Ratinng'][score.index[i]],
                     "genre":movie_df['genre'][score.index[i]],
                     "act":movie_df['act'][score.index[i]]
                    }
        lst.append(Movie_title)
    return lst
    
```

```python
movie_recom("3 Idiots",10)
```

```python
df=pd.DataFrame(movie_recom("3 Idiots",5))
```
